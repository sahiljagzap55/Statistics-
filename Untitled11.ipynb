```python
# Exploratory Data Analysis (EDA)
# Practical: Student Performance Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --------------------------------
# 1. Create a Dataset
# --------------------------------

data = {
    "Student_ID": range(1, 16),
    "Study_Hours": [
        2, 3, 5, 4, 6, 7, 1, 8, 5, 3, 6, 4, 9, 2, 7
    ],
    "Attendance": [
        65, 70, 85, 75, 90, 95, 50, 98, 82, 68, 88, 78, 99, 60, 92
    ],
    "Previous_Score": [
        55, 60, 72, 65, 80, 85, 40, 90, 75, 58, 82, 68, 95, 50, 88
    ],
    "Final_Score": [
        50, 58, 75, 65, 82, 88, 35, 95, 78, 55, 85, 70, 98, 45, 90
    ]
}

df = pd.DataFrame(data)

# Introduce a missing value for demonstration
df.loc[3, "Previous_Score"] = np.nan

# Save the dataset
df.to_csv("student_performance.csv", index=False)

print("\n===== ORIGINAL DATASET =====")
print(df)

# --------------------------------
# 2. Data Processing
# --------------------------------

print("\n===== DATASET INFORMATION =====")
print(df.info())

print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

# Fill missing values with the column mean
df["Previous_Score"] = df["Previous_Score"].fillna(
    df["Previous_Score"].mean()
)

print("\n===== DATA AFTER CLEANING =====")
print(df)

# --------------------------------
# 3. Statistical Analysis
# --------------------------------

print("\n===== STATISTICAL SUMMARY =====")
print(df.describe())

# --------------------------------
# 4. Feature Selection
# --------------------------------

print("\n===== CORRELATION MATRIX =====")

correlation = df[
    ["Study_Hours", "Attendance", "Previous_Score", "Final_Score"]
].corr()

print(correlation)

# --------------------------------
# 5. Data Visualization
# --------------------------------

sns.set_theme(style="whitegrid")

# Graph 1: Study Hours vs Final Score
plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Final_Score",
    s=100
)

plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.tight_layout()
plt.savefig("study_hours_vs_score.png")
plt.show()

# Graph 2: Correlation Heatmap
plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.savefig("correlation_heatmap.png")
plt.show()

# Graph 3: Final Score Distribution
plt.figure(figsize=(7, 5))

sns.histplot(
    data=df,
    x="Final_Score",
    bins=6,
    kde=True
)

plt.title("Distribution of Final Scores")
plt.xlabel("Final Score")
plt.tight_layout()
plt.savefig("final_score_distribution.png")
plt.show()

# --------------------------------
# 6. Final Result
# --------------------------------

print("\n===== EDA COMPLETED SUCCESSFULLY =====")
print("Dataset created: student_performance.csv")
print("Missing values handled successfully.")
print("Statistical analysis completed.")
print("Feature correlation calculated.")
print("Three visualization graphs generated.")
```
